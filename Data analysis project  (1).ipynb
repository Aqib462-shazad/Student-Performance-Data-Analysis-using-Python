# Data cleaning 

```python
df.isnull().sum()
```

```python
columns_to_fill = ['EthnicGroup', 'ParentEduc', 'TestPrep', 'ParentMaritalStatus', 
                   'PracticeSport', 'IsFirstChild', 'TransportMeans', ]

for column in columns_to_fill:
    df[column].fillna(df[column].mode()[0], inplace=True)
```

```python
df = df.drop("Unnamed: 0", axis = 1)
print(df.head())
```



# For Regularly Practice Sport 

```python
females_regularly = df[(df["Gender"] == "female") & (df["PracticeSport"] == "regularly")]
males_regularly = df[(df["Gender"] == "male") & (df["PracticeSport"] == "regularly")]
count_females_regularly = females_regularly.shape[0]
count_males_regularly = males_regularly.shape[0]
print(f"Females practicing sports regularly: {count_females_regularly}")
print(f"Males practicing sports regularly: {count_males_regularly}")
categories = ['Females', 'Males']
counts = [count_females_regularly, count_males_regularly]
plt.bar(categories, counts, color=['pink', 'blue'])
plt.title('Regular Practice of Sports by Gender')
plt.ylabel('Count')
plt.xlabel('Gender')
#plt.ylim(0, max(counts) + 1)  # Adjusting y-axis for better visibility
plt.show()
```

# For Sometime Practice Sport

```python
females_sometime = df[(df["Gender"] == "female") & (df["PracticeSport"] == "sometimes")]
males_sometime = df[(df["Gender"] == "male") & (df["PracticeSport"] == "sometimes")]
count_females_sometime = females_sometime.shape[0]
count_males_sometime = males_sometime.shape[0]
print(f"Females practicing sports sometime: {count_females_sometime}")
print(f"Males practicing sports sometime: {count_males_sometime}")
categories = ['Females', 'Males']
counts = [count_females_sometime, count_males_sometime]
plt.bar(categories, counts, color=['pink', 'blue'])
plt.title('Sometime Practice of Sports by Gender')
plt.ylabel('Count')
plt.xlabel('Gender')
#plt.ylim(0, max(counts) + 1)  # Adjusting y-axis for better visibility
plt.show()
```
